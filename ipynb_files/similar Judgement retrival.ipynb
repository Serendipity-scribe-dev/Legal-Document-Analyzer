```python
import os
import pandas as pd

# Path to judgment files
folder_path = "Object_casedocs"

# Read all text files into a DataFrame
judgments = []
for file in os.listdir(folder_path):
    if file.endswith(".txt"):
        with open(os.path.join(folder_path, file), "r", encoding="utf-8") as f:
            judgments.append({"filename": file, "text": f.read()})

df = pd.DataFrame(judgments)
print(df.head())  # Display the first few rows
```

```python
import pandas as pd
from sentence_transformers import SentenceTransformer

# Load pre-trained model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Convert case texts into embeddings
case_embeddings = model.encode(df["text"].tolist(), convert_to_tensor=True)

# Function to find similar cases using cosine similarity
import torch

def find_similar_cases(query, top_n=5):
    query_embedding = model.encode([query], convert_to_tensor=True)
    similarity_scores = torch.nn.functional.cosine_similarity(query_embedding, case_embeddings)
    top_indices = torch.topk(similarity_scores, top_n).indices.tolist()
    
    print("\n🔍 **Top Similar Cases** 🔍")
    for i in top_indices:
        print(f"\n📜 **Judgment File:** {df.iloc[i]['filename']}")
        print("📖 **Snippet of Case Content:**\n")
        print(df.iloc[i]['text'][:1000])  # Display only a snippet to keep it readable
        print("\n" + "="*100 + "\n")

query = "A group of men murdered a girl"
find_similar_cases(query)
```

```python
import os
import pandas as pd
import torch
from sentence_transformers import SentenceTransformer

# 📂 Load judgment text files into a DataFrame
judgment_folder = "Object_casedocs"  # 🔹 Update with your folder path
file_names = []
file_contents = []

for file in os.listdir(judgment_folder):
    if file.endswith(".txt"):  
        file_names.append(file)
        with open(os.path.join(judgment_folder, file), "r", encoding="utf-8") as f:
            file_contents.append(f.read())  

# 🗂️ Create DataFrame
df = pd.DataFrame({"filename": file_names, "text": file_contents})

# 🔹 Load pre-trained Sentence Transformer model
model = SentenceTransformer('all-MiniLM-L6-v2')

# 🔢 Convert case texts into embeddings
case_embeddings = model.encode(df["text"].tolist(), convert_to_tensor=True)

# 🔍 Function to find similar cases
def find_similar_cases(query, top_n=5):
    query_embedding = model.encode([query], convert_to_tensor=True)
    similarity_scores = torch.nn.functional.cosine_similarity(query_embedding, case_embeddings)
    top_indices = torch.topk(similarity_scores, top_n).indices.tolist()  # Get top N cases
    
    print("\n🔍 **Top Similar Cases** 🔍")
    for i in top_indices:
        print(f"\n📜 **Judgment File:** {df.iloc[i]['filename']}")
        print("📖 **Snippet of Case Content:**\n")
        print(df.iloc[i]['text'][:1000])  # Display first 1000 characters
        print("\n" + "="*100 + "\n")

# 🏛️ Example Query
query = "A group of men murdered a girl"
find_similar_cases(query)
```

```python
from sentence_transformers import SentenceTransformer, util
import pandas as pd
import os

# Load pre-trained sentence transformer model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Load judgments dataset
judgment_folder = "Object_casedocs"
judgment_files = os.listdir(judgment_folder)

# Read and encode all judgments
judgments = {}
judgment_embeddings = {}
for file in judgment_files:
    with open(os.path.join(judgment_folder, file), "r", encoding="utf-8") as f:
        content = f.read()
        judgments[file] = content
        judgment_embeddings[file] = model.encode(content, convert_to_tensor=True)

def find_similar_cases(query, top_k=5):
    query_embedding = model.encode(query, convert_to_tensor=True)
    
    similarities = []
    for file, embedding in judgment_embeddings.items():
        score = util.pytorch_cos_sim(query_embedding, embedding).item()
        similarities.append((file, score))
    
    # Sort by similarity score in descending order
    similarities.sort(key=lambda x: x[1], reverse=True)
    
    # Get top k results
    top_cases = similarities[:top_k]
    
    # Display results
    print("\n🔍 **Top Similar Cases** 🔍")
    for case, score in top_cases:
        snippet = " ".join(judgments[case].split()[:50])  # First 50 words as snippet
        print(f"\n📜 **Judgment File:** {case}")
        print(f"📖 **Snippet of Case Content:** {snippet}...")
        print(f"🔹 **Similarity Score:** {score:.4f}")
        print("="*100)

# Example usage
query = "A group of men murdered a girl"
find_similar_cases(query)
```
