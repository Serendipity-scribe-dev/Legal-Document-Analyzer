```python
import pandas as pd

# File paths
ipc_sections_path = "ipc_sections.csv"

# Load datasets
ipc_sections_df = pd.read_csv(ipc_sections_path)

# Display first few rows of each dataset
ipc_sections_df.head()
```

```python
import re

# Function to clean IPC descriptions (remove newlines and extra spaces)
def clean_text(text):
    text = re.sub(r'\n+', ' ', str(text))  # Replace newlines with spaces
    text = re.sub(r'\s+', ' ', text).strip()  # Remove extra spaces
    return text

# Apply cleaning to both datasets
ipc_sections_df["Description"] = ipc_sections_df["Description"].apply(clean_text)
ipc_sections_df["Offense"] = ipc_sections_df["Offense"].apply(clean_text)
ipc_sections_df["Punishment"] = ipc_sections_df["Punishment"].apply(clean_text)

# Extract IPC section numbers from the "Section" column
ipc_sections_df["Section_Number"] = ipc_sections_df["Section"].apply(lambda x: re.findall(r'\d+', str(x))[0] if re.findall(r'\d+', str(x)) else None)

# Display cleaned dataset samples
ipc_sections_df.head()
```

```python
from sentence_transformers import SentenceTransformer

# Load the SBERT model
model = SentenceTransformer('all-mpnet-base-v2')

# Convert IPC descriptions into embeddings
ipc_embeddings = model.encode(ipc_sections_df["Description"].tolist(), convert_to_tensor=True)

# Store embeddings in the DataFrame
ipc_sections_df["Embedding"] = ipc_embeddings.tolist()

# Check the first few embeddings
ipc_sections_df.head()
```

```python
import numpy as np

# Generate embeddings
ipc_embeddings = model.encode(ipc_sections_df["Description"].tolist(), convert_to_tensor=True)
ipc_embeddings_np = np.array([embedding.cpu().numpy() for embedding in ipc_embeddings])
```

```python
import faiss
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Compute cosine similarity matrix
cosine_sim_matrix = cosine_similarity(ipc_embeddings_np)

# Print similarity matrix (optional)
print("Cosine Similarity Matrix:")
print(cosine_sim_matrix)

# Create FAISS index for nearest neighbor search
d = ipc_embeddings_np.shape[1]
ipc_index = faiss.IndexFlatIP(d)  # Inner Product index for cosine similarity
faiss.normalize_L2(ipc_embeddings_np)  # Normalize embeddings to unit vectors
ipc_index.add(ipc_embeddings_np)

# Function to find multiple relevant IPC sections using cosine similarity
def find_related_ipc(query, k=5):
    query_embedding = model.encode([query]).astype("float32")
    faiss.normalize_L2(query_embedding)  # Normalize query embedding

    # Retrieve top-k matches based on cosine similarity
    scores, indices = ipc_index.search(query_embedding, k=k)

    result_str = ""
    for score, idx in zip(scores[0], indices[0]):  # Iterate over matches
        section_info = ipc_sections_df.iloc[idx]
        result_str += f"IPC Section: {section_info['Section']}\n"
        result_str += f"Description: {section_info['Description']}\n"
        result_str += f"Offense: {section_info.get('Offense', 'N/A')}\n"
        result_str += f"Punishment: {section_info.get('Punishment', 'N/A')}\n"
        result_str += f"🔹 Similarity Score: {score:.4f}\n"
        result_str += "-" * 50 + "\n"

    return result_str


# Example Usage
query = "theft and property damage"
print(find_related_ipc(query, k=5))  # Fetch top 5 matches
```

```python
# Test IPC section retrieval
case_scenario = "A person was attacked with a sharp weapon"
ipc_result = find_related_ipc(case_scenario)
print("🔎 IPC Section Match:")
print(ipc_result)

```
